* Diagnosis
    - Alzheimer’s disease is diagnosed using a combination of clinical evaluations, cognitive testing, laboratory tests, brain imaging, and biological biomarkers. Clinicians evaluate changes in memory, problem solving, attention, language, behavior, and the ability to perform everyday activities while also testing for other conditions that could cause similar symptoms. Brain imaging techniques such as MRI and CT can identify structural abnormalities and help rule out other causes, while PET imaging can detect amyloid plaques or tau tangles associated with Alzheimer’s disease. Cerebrospinal fluid obtained through a lumbar puncture can also be tested for biomarkers including amyloid beta, tau, and phosphorylated tau (p-tau).   Blood based biomarkers are becoming increasingly important; in 2025, the FDA cleared the first blood test used to aid Alzheimer’s diagnosis, which measures the ratio of plasma p-tau217 to Aβ1-42 in symptomatic adults age 55 and older.   These advances are making it possible to identify the biological changes associated with Alzheimer’s more accurately and potentially earlier in the disease process.

* Standard of care treatments (& reimbursement) 
    - There is currently no cure for Alzheimer’s disease, but several treatments are available to manage symptoms or slow disease progression in certain patients. Medications such as donepezil, rivastigmine, galantamine, and memantine can help manage cognitive symptoms by affecting neurotransmitter systems involved in memory and thinking. Newer disease modifying treatments include lecanemab (Leqembi) and donanemab (Kisunla), which are monoclonal antibodies that target amyloid beta and reduce amyloid plaques in the brain. These treatments are intended for patients in the early stages of Alzheimer’s disease, such as mild cognitive impairment or mild dementia, rather than patients with advanced disease. Medicare Part B covers FDA traditionally approved anti amyloid treatments such as lecanemab and donanemab for eligible patients when CMS coverage requirements are met, including participation by the treating clinician in a qualifying study or registry. Although these treatments do not cure or reverse Alzheimer’s disease, newer disease modifying therapies can slow cognitive and functional decline in some patients with early Alzheimer’s disease.

* Disease progression & prognosis
    - Alzheimer’s disease is progressive, meaning that cognitive and functional abilities generally worsen over time as the disease advances. Biological changes in the brain can begin a decade or more before noticeable symptoms appear, followed by stages that can progress from mild cognitive impairment to mild, moderate, and eventually severe dementia. During mild Alzheimer’s disease, individuals may experience memory problems and difficulty completing everyday tasks, while moderate disease can involve worsening confusion, difficulty recognizing family and friends, and behavioral changes. In severe Alzheimer’s disease, individuals can lose the ability to communicate and become completely dependent on others for care. The rate of progression and prognosis vary considerably between individuals; among people age 65 and older, survival averages approximately four to eight years following an Alzheimer’s dementia diagnosis, although some individuals may live as long as 20 years. As Alzheimer’s progresses, increasing neuronal dysfunction and death cause brain tissue loss and contribute to the worsening cognitive and functional impairment associated with the disease.

* Continuum of care providers
    - Alzheimer’s disease requires a continuum of care involving multiple healthcare professionals as a patient’s needs change throughout disease progression. Primary care physicians often provide ongoing medical care and may refer patients to specialists such as neurologists, geriatricians, psychiatrists, or memory-disorders clinics for further evaluation and treatment.   Other professionals, including nurses, neuropsychologists, social workers, pharmacists, and physical, occupational, or speech therapists, may help manage cognitive, behavioral, physical, and everyday functional difficulties. Family members and other caregivers also become increasingly important as patients lose the ability to independently perform activities of daily living. As the disease advances, patients may require professional home care services, assisted living, memory care, or nursing home care when their needs can no longer be safely managed independently. Coordinating these different providers and caregivers throughout disease progression is important for maintaining the patient’s health, safety, and quality of life.

* Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)
    - Alzheimer’s disease causes progressive changes in the brain, with early damage commonly occurring in the hippocampus and entorhinal cortex, regions that are important for learning and memory. As the disease progresses, neurons lose connections, stop functioning properly, and eventually die, causing affected regions of the brain to shrink. Two major molecular features of Alzheimer’s disease are extracellular amyloid beta plaques and intracellular neurofibrillary tangles composed of abnormal tau protein. In healthy neurons, tau normally binds to and stabilizes microtubules, which help transport nutrients and other molecules through the cell. In Alzheimer’s disease, abnormal chemical changes to tau cause it to detach from microtubules and associate with other tau molecules, eventually forming neurofibrillary tangles that interfere with the neuron’s transport and communication systems. Phosphorylated tau (p-tau), the biomarker examined in this project, is associated with this abnormal tau pathology and provides a measurable connection between molecular changes in Alzheimer’s disease and the cognitive decline that occurs as neuronal damage spreads through the brain.

* Clinical Trials/next-gen therapies
    - Alzheimer’s disease research is increasingly focused on developing therapies that target the biological mechanisms responsible for disease progression rather than only managing symptoms. Although recently developed disease modifying treatments primarily target amyloid beta, current clinical trials are investigating a much wider range of targets, including tau, neuroinflammation, metabolism, vascular function, and other mechanisms involved in neuronal damage. Tau directed therapies are particularly relevant because abnormal tau accumulation is strongly associated with Alzheimer’s pathology and is the focus of several ongoing experimental treatments. For example, the NIA supported Alzheimer’s Disease Tau Platform is evaluating tau directed therapies both alone and in combination with anti amyloid treatment to determine whether they can slow disease progression. Other next generation approaches under investigation include small molecule drugs, vaccines, immunotherapies, and even gene therapies that target different components of Alzheimer’s disease biology. The diversity of these clinical trials suggests that future Alzheimer’s treatment may involve targeting multiple disease mechanisms rather than relying on a single therapeutic approach.
