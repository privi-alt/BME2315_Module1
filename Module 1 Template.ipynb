* Diagnosis
    - Alzheimer’s disease is diagnosed using a combination of clinical evaluations, cognitive testing, laboratory tests, brain imaging, and biological biomarkers. Clinicians evaluate changes in memory, problem solving, attention, language, behavior, and the ability to perform everyday activities while also testing for other conditions that could cause similar symptoms. Brain imaging techniques such as MRI and CT can identify structural abnormalities and help rule out other causes, while PET imaging can detect amyloid plaques or tau tangles associated with Alzheimer’s disease. Cerebrospinal fluid obtained through a lumbar puncture can also be tested for biomarkers including amyloid beta, tau, and phosphorylated tau (p-tau).   Blood based biomarkers are becoming increasingly important; in 2025, the FDA cleared the first blood test used to aid Alzheimer’s diagnosis, which measures the ratio of plasma p-tau217 to Aβ1-42 in symptomatic adults age 55 and older.   These advances are making it possible to identify the biological changes associated with Alzheimer’s more accurately and potentially earlier in the disease process.

Newer disease modifying treatments include lecanemab (Leqembi) and donanemab (Kisunla), which are monoclonal antibodies that target amyloid beta and reduce amyloid plaques in the brain. These treatments are intended for patients in the early stages of Alzheimer’s disease, such as mild cognitive impairment or mild dementia, rather than patients with advanced disease. Medicare Part B covers FDA traditionally approved anti amyloid treatments such as lecanemab and donanemab for eligible patients when CMS coverage requirements are met, including participation by the treating clinician in a qualifying study or registry. Although these treatments do not cure or reverse Alzheimer’s disease, newer disease modifying therapies can slow cognitive and functional decline in some patients with early Alzheimer’s disease.

* Disease progression & prognosis
    - Alzheimer’s disease is progressive, meaning that cognitive and functional abilities generally worsen over time as the disease advances. Biological changes in the brain can begin a decade or more before noticeable symptoms appear, followed by stages that can progress from mild cognitive impairment to mild, moderate, and eventually severe dementia. During mild Alzheimer’s disease, individuals may experience memory problems and difficulty completing everyday tasks, while moderate disease can involve worsening confusion, difficulty recognizing family and friends, and behavioral changes. In severe Alzheimer’s disease, individuals can lose the ability to communicate and become completely dependent on others for care. The rate of progression and prognosis vary considerably between individuals; among people age 65 and older, survival averages approximately four to eight years following an Alzheimer’s dementia diagnosis, although some individuals may live as long as 20 years. As Alzheimer’s progresses, increasing neuronal dysfunction and death cause brain tissue loss and contribute to the worsening cognitive and functional impairment associated with the disease.

* Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)
    - Alzheimer’s disease causes progressive changes in the brain, with early damage commonly occurring in the hippocampus and entorhinal cortex, regions that are important for learning and memory. As the disease progresses, neurons lose connections, stop functioning properly, and eventually die, causing affected regions of the brain to shrink. Two major molecular features of Alzheimer’s disease are extracellular amyloid beta plaques and intracellular neurofibrillary tangles composed of abnormal tau protein. In healthy neurons, tau normally binds to and stabilizes microtubules, which help transport nutrients and other molecules through the cell. In Alzheimer’s disease, abnormal chemical changes to tau cause it to detach from microtubules and associate with other tau molecules, eventually forming neurofibrillary tangles that interfere with the neuron’s transport and communication systems. Phosphorylated tau (p-tau), the biomarker examined in this project, is associated with this abnormal tau pathology and provides a measurable connection between molecular changes in Alzheimer’s disease and the cognitive decline that occurs as neuronal damage spreads through the brain.

## Data Analysis

Three analyses were performed to investigate the relationships between pTAU levels, sex, cognitive status, and age at death. Independent-samples t-tests were used to compare mean pTAU levels between groups, while linear regression was used to investigate the relationship between age at death and pTAU.


```python
from patient import Patient

import matplotlib.pyplot as plt
import statistics
from scipy import stats

# Load patient data from the CSV file
Patient.instantiate_from_csv("Metadata and Protein Data for Module 1.csv")
```

### pTAU Levels by Sex in Patients with Dementia

To investigate whether pTAU levels differed by sex, patients with dementia were separated into female and male groups. The mean and standard deviation of pTAU were calculated for each group and displayed using a bar graph with standard deviation error bars. An independent-samples t-test was used to determine whether the difference between the groups was statistically significant.

```python
# Filter female and male patients with dementia
female_dementia = Patient.filter("Female", "Dementia")
male_dementia = Patient.filter("Male", "Dementia")

# Calculate mean and standard deviation of pTAU by sex
female_ptau = [patient.ptau for patient in female_dementia]
male_ptau = [patient.ptau for patient in male_dementia]

female_mean = statistics.mean(female_ptau)
male_mean = statistics.mean(male_ptau)
female_sd = statistics.stdev(female_ptau)
male_sd = statistics.stdev(male_ptau)

groups = ["Female", "Male"]
means = [female_mean, male_mean]
standard_deviations = [female_sd, male_sd]

# Compare female and male pTAU levels using an independent t-test
t_statistic, p_value = stats.ttest_ind(female_ptau, male_ptau)

print("Female vs. Male pTAU t-statistic:", t_statistic)
print("Female vs. Male pTAU p-value:", p_value)

# Create bar graph
plt.figure()
plt.bar(groups, means, yerr=standard_deviations, capsize=5)

plt.xlabel("Sex")
plt.ylabel("Mean pTAU (pg/ug)")
plt.title("Mean pTAU in Dementia Patients by Sex")

plt.show()
```

The difference in pTAU between female and male patients with dementia was not statistically significant (t = 0.616, p = 0.542). Because the p-value was greater than 0.05, the data did not provide sufficient evidence of a difference in mean pTAU levels between female and male patients with dementia.

### Relationship Between Age at Death and pTAU

To investigate whether age at death was associated with pTAU levels, age at death and pTAU were plotted using a scatter plot. Linear regression was performed to evaluate the strength and statistical significance of the relationship.

```python
ages = [patient.age_at_death for patient in Patient.all_patients]
ptau_values = [patient.ptau for patient in Patient.all_patients]

# Calculate linear regression for age at death vs. pTAU
slope, intercept, r_value, p_value, std_err = stats.linregress(ages, ptau_values)

regression_line = [slope * age + intercept for age in ages]

print("Age vs. pTAU slope:", slope)
print("Age vs. pTAU intercept:", intercept)
print("Age vs. pTAU R-squared:", r_value ** 2)
print("Age vs. pTAU p-value:", p_value)

# Create scatter plot with regression line
plt.figure()
plt.scatter(ages, ptau_values)
plt.plot(ages, regression_line)

plt.xlabel("Age at Death")
plt.ylabel("pTAU (pg/ug)")
plt.title("Age at Death vs. pTAU with Linear Regression")

plt.show()
```

The regression had a slope of -0.039 pg/µg per year, indicating a small negative relationship between age at death and pTAU. However, the R² value was 0.015, meaning that age at death explained approximately 1.5% of the variation in pTAU levels. The relationship was not statistically significant (p = 0.271). Therefore, this dataset did not provide sufficient evidence of a linear relationship between age at death and pTAU.

### pTAU Levels by Cognitive Status

To investigate the relationship between pTAU and cognitive status, patients were separated into "Dementia" and "No dementia" groups. Mean pTAU and standard deviation were calculated for each group and displayed using a bar graph with standard deviation error bars. An independent-samples t-test was used to compare mean pTAU between the two cognitive-status groups.

```python
# Filter patients by cognitive status
dementia = Patient.filter("any", "Dementia")
no_dementia = Patient.filter("any", "No dementia")

# Get pTAU values for each cognitive status group
dementia_ptau = [patient.ptau for patient in dementia]
no_dementia_ptau = [patient.ptau for patient in no_dementia]

# Compare groups using an independent t-test
t_statistic, p_value = stats.ttest_ind(dementia_ptau, no_dementia_ptau)

print("Dementia vs. No dementia pTAU t-statistic:", t_statistic)
print("Dementia vs. No dementia pTAU p-value:", p_value)

# Calculate mean and standard deviation of pTAU for each group
dementia_mean = statistics.mean(dementia_ptau)
no_dementia_mean = statistics.mean(no_dementia_ptau)

dementia_sd = statistics.stdev(dementia_ptau)
no_dementia_sd = statistics.stdev(no_dementia_ptau)

# Create bar graph
groups = ["Dementia", "No dementia"]
means = [dementia_mean, no_dementia_mean]
standard_deviations = [dementia_sd, no_dementia_sd]

plt.figure()
plt.bar(groups, means, yerr=standard_deviations, capsize=5)

plt.xlabel("Cognitive Status")
plt.ylabel("Mean pTAU (pg/ug)")
plt.title("Mean pTAU by Cognitive Status")

plt.show()
```

Mean pTAU was higher in the dementia group than in the no-dementia group; however, the difference was not statistically significant (t = 1.611, p = 0.111). Because the p-value was greater than 0.05, the data did not provide sufficient evidence of a difference in mean pTAU between the two cognitive-status groups.

## Limitations and Future Work

Several limitations should be considered when interpreting our results. First, the dataset contained only 84 patients, which limits the statistical power of the analysis and may make smaller relationships more difficult to detect. There was also substantial variability in pTAU levels within the groups, which can make differences between group means more difficult to distinguish statistically. Additionally, our analyses examined sex, cognitive status, and age separately and did not control for other potentially important variables available in the dataset, such as APOE genotype, Braak stage, and Thal phase. Future work could use a larger dataset and include additional variables to better understand the factors associated with pTAU. For example, patients could be separated by APOE genotype to investigate whether the relationship between sex and pTAU differs based on genetic risk. More advanced statistical methods, such as multiple regression, could examine several patient characteristics simultaneously rather than analyzing each variable separately. Future studies could also investigate specific p-tau biomarkers and use longitudinal measurements to determine how p-tau changes as Alzheimer's disease progresses over time.
