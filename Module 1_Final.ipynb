* Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)
    - Alzheimer’s disease causes progressive changes in the brain, with early damage commonly occurring in the hippocampus and entorhinal cortex, regions that are important for learning and memory. As the disease progresses, neurons lose connections, stop functioning properly, and eventually die, causing affected regions of the brain to shrink. Two major molecular features of Alzheimer’s disease are extracellular amyloid beta plaques and intracellular neurofibrillary tangles composed of abnormal tau protein. In healthy neurons, tau normally binds to and stabilizes microtubules, which help transport nutrients and other molecules through the cell. In Alzheimer’s disease, abnormal chemical changes to tau cause it to detach from microtubules and associate with other tau molecules, eventually forming neurofibrillary tangles that interfere with the neuron’s transport and communication systems. Phosphorylated tau (p-tau), the biomarker examined in this project, is associated with this abnormal tau pathology and provides a measurable connection between molecular changes in Alzheimer’s disease and the cognitive decline that occurs as neuronal damage spreads through the brain.

## Data Analysis

Four analyses were performed to investigate differences and relationships in pTAU levels based on sex, cognitive status, and age at death. Independent-samples t-tests were used to compare mean pTAU levels between female and male patients with dementia and between patients with dementia and patients without dementia. A one-way ANOVA was used to compare mean pTAU levels across four groups defined by sex and cognitive status. Linear regression was used to evaluate the relationship between age at death and pTAU as continuous variables.


```python
# Import the csv module so the patient data can be read from the CSV file
import csv

# Define a Patient class to store information for each donor
class Patient:

    # List that will store every Patient object created
    all_patients = []

    # Initialize the attributes for each patient
    def __init__(
        self,
        donor_id,
        age_at_death,
        sex,
        apoe_genotype,
        cognitive_status,
        abeta42,
        ptau
    ):
        self.donor_id = donor_id
        self.age_at_death = age_at_death
        self.sex = sex
        self.apoe_genotype = apoe_genotype
        self.cognitive_status = cognitive_status
        self.abeta42 = abeta42
        self.ptau = ptau

        # Add each new patient to the list of all patients
        Patient.all_patients.append(self)

    # Define how a patient will appear when printed
    def __repr__(self):
        return (
            f"{self.donor_id} | Age: {self.age_at_death} | "
            f"{self.sex} | {self.cognitive_status} | pTAU: {self.ptau}"
        )

    # Read the CSV file and create one Patient object for each donor
    @classmethod
    def instantiate_from_csv(cls, filename):
        with open(filename, "r") as f:
            reader = csv.DictReader(f)

            for row in reader:
                Patient(
                    donor_id=row["Donor ID"],
                    age_at_death=int(row["Age at Death"]),
                    sex=row["Sex"],
                    apoe_genotype=row["APOE Genotype"],
                    cognitive_status=row["Cognitive Status"],
                    abeta42=float(row["ABeta42 pg/ug"]),
                    ptau=float(row["pTAU pg/ug"])
                )

    # Filter patients using both sex and cognitive status
    @classmethod
    def filter(cls, sex, cognitive_status):
        filtered_patients = []

        for patient in cls.all_patients:
            if (
                patient.sex == sex
                and patient.cognitive_status == cognitive_status
            ):
                filtered_patients.append(patient)

        return filtered_patients


# Import libraries for graphs and statistical analysis
import matplotlib.pyplot as plt
import statistics
from scipy import stats

# Load all patient data from the course CSV file
Patient.instantiate_from_csv(
    "Metadata and Protein Data for Module 1.csv"
)

# Check that all 84 donors were loaded correctly
print("Number of patients loaded:", len(Patient.all_patients))
```

### pTAU Levels by Sex in Patients with Dementia

To investigate whether pTAU levels differed by sex, patients with dementia were separated into female and male groups. The mean and standard deviation of pTAU were calculated for each group and displayed using a bar graph with standard deviation error bars. An independent-samples t-test was used to determine whether the difference between the groups was statistically significant.

```python
# Filter female and male patients with dementia
female_dementia = Patient.filter("Female", "Dementia")
male_dementia = Patient.filter("Male", "Dementia")

# Calculate mean and standard deviation of pTAU by sex
female_ptau = [patient.ptau for patient in female_dementia]
male_ptau = [patient.ptau for patient in male_dementia]

female_mean = statistics.mean(female_ptau)
male_mean = statistics.mean(male_ptau)
female_sd = statistics.stdev(female_ptau)
male_sd = statistics.stdev(male_ptau)

groups = ["Female", "Male"]
means = [female_mean, male_mean]
standard_deviations = [female_sd, male_sd]

# Compare female and male pTAU levels using an independent t-test
t_statistic, p_value = stats.ttest_ind(female_ptau, male_ptau)

print("Female vs. Male pTAU t-statistic:", t_statistic)
print("Female vs. Male pTAU p-value:", p_value)

# Create bar graph
plt.figure()
plt.bar(groups, means, yerr=standard_deviations, capsize=5)

plt.xlabel("Sex")
plt.ylabel("Mean pTAU (pg/ug)")
plt.title("Mean pTAU in Dementia Patients by Sex")

plt.show()

```

The difference in pTAU between female and male patients with dementia was not statistically significant (t = 0.616, p = 0.542). Because the p-value was greater than 0.05, the data did not provide sufficient evidence of a difference in mean pTAU levels between female and male patients with dementia.

### Relationship Between Age at Death and pTAU

To investigate whether age at death was associated with pTAU levels, age at death and pTAU were plotted using a scatter plot. Linear regression was performed to evaluate the strength and statistical significance of the relationship.

```python
# Get age at death and pTAU values for all patients
ages = [patient.age_at_death for patient in Patient.all_patients]
ptau_values = [patient.ptau for patient in Patient.all_patients]

# Perform linear regression to test the relationship
# between age at death and pTAU
slope, intercept, r_value, p_value, std_err = stats.linregress(
    ages,
    ptau_values
)

# Sort the ages so the regression line is drawn correctly
sorted_ages = sorted(ages)

# Calculate predicted pTAU values using the regression equation
regression_line = [
    slope * age + intercept
    for age in sorted_ages
]

# Print the linear regression results
print("Age vs. pTAU slope:", slope)
print("Age vs. pTAU intercept:", intercept)
print("Age vs. pTAU R:", r_value)
print("Age vs. pTAU R-squared:", r_value ** 2)
print("Age vs. pTAU p-value:", p_value)
print("Age vs. pTAU standard error:", std_err)

# Create scatter plot showing the patient data
plt.figure()
plt.scatter(ages, ptau_values)

# Add the linear regression line
plt.plot(sorted_ages, regression_line)

# Label the graph
plt.xlabel("Age at Death")
plt.ylabel("pTAU (pg/ug)")
plt.title("Age at Death vs. pTAU with Linear Regression")

plt.show()
```

The linear regression showed a weak negative relationship between age at death and pTAU, with a slope of approximately -0.039 pg/ug per year. However, the R² value was only 0.015, meaning age at death explained approximately 1.5% of the variation in pTAU. The relationship was not statistically significant (p = 0.271). Therefore, this dataset did not provide sufficient evidence of a linear relationship between age at death and pTAU.

### pTAU Levels by Cognitive Status

To investigate the relationship between pTAU and cognitive status, patients were separated into "Dementia" and "No dementia" groups. Mean pTAU and standard deviation were calculated for each group and displayed using a bar graph with standard deviation error bars. An independent-samples t-test was used to compare mean pTAU between the two cognitive-status groups.

```python
# Filter all patients based only on cognitive status
# We use list comprehensions here because Patient.filter()
# requires both sex and cognitive status to match exactly.

# Get all patients with dementia
dementia = [
    patient for patient in Patient.all_patients
    if patient.cognitive_status == "Dementia"
]

# Get all patients without dementia
no_dementia = [
    patient for patient in Patient.all_patients
    if patient.cognitive_status == "No dementia"
]

# Get pTAU values for each cognitive status group
dementia_ptau = [patient.ptau for patient in dementia]
no_dementia_ptau = [patient.ptau for patient in no_dementia]

# Compare groups using an independent t-test
t_statistic, p_value = stats.ttest_ind(dementia_ptau, no_dementia_ptau)

print("Dementia vs. No dementia pTAU t-statistic:", t_statistic)
print("Dementia vs. No dementia pTAU p-value:", p_value)

# Calculate mean and standard deviation of pTAU for each group
dementia_mean = statistics.mean(dementia_ptau)
no_dementia_mean = statistics.mean(no_dementia_ptau)

dementia_sd = statistics.stdev(dementia_ptau)
no_dementia_sd = statistics.stdev(no_dementia_ptau)

# Print descriptive statistics for each cognitive-status group
print("Dementia mean pTAU:", dementia_mean)
print("Dementia pTAU standard deviation:", dementia_sd)
print("No dementia mean pTAU:", no_dementia_mean)
print("No dementia pTAU standard deviation:", no_dementia_sd)

# Create bar graph
groups = ["Dementia", "No dementia"]
means = [dementia_mean, no_dementia_mean]
standard_deviations = [dementia_sd, no_dementia_sd]

plt.figure()
plt.bar(groups, means, yerr=standard_deviations, capsize=5)

plt.xlabel("Cognitive Status")
plt.ylabel("Mean pTAU (pg/ug)")
plt.title("Mean pTAU by Cognitive Status")

plt.show()
```

Mean pTAU was higher in the dementia group than in the no-dementia group; however, the difference was not statistically significant (t = 1.611, p = 0.111). Because the p-value was greater than 0.05, the data did not provide sufficient evidence of a difference in mean pTAU between the two cognitive-status groups.

### pTAU Levels by Sex and Cognitive Status

To examine whether pTAU levels differed when considering both sex and cognitive status, patients were divided into four groups: female patients with dementia, male patients with dementia, female patients without dementia, and male patients without dementia. A one-way ANOVA was used to test whether mean pTAU differed among the four groups.

```python
# Create four groups based on sex and cognitive status
female_dementia = Patient.filter("Female", "Dementia")
male_dementia = Patient.filter("Male", "Dementia")
female_no_dementia = Patient.filter("Female", "No dementia")
male_no_dementia = Patient.filter("Male", "No dementia")

# Extract pTAU values from each group
female_dementia_ptau = [
    patient.ptau for patient in female_dementia
]

male_dementia_ptau = [
    patient.ptau for patient in male_dementia
]

female_no_dementia_ptau = [
    patient.ptau for patient in female_no_dementia
]

male_no_dementia_ptau = [
    patient.ptau for patient in male_no_dementia
]

# Run a one-way ANOVA to compare mean pTAU across all four groups
f_statistic, anova_p_value = stats.f_oneway(
    female_dementia_ptau,
    male_dementia_ptau,
    female_no_dementia_ptau,
    male_no_dementia_ptau
)

# Print the ANOVA results
print("ANOVA F-statistic:", f_statistic)
print("ANOVA p-value:", anova_p_value)

# Interpret the ANOVA using a significance level of 0.05
if anova_p_value < 0.05:
    print("The difference among the groups is statistically significant.")
else:
    print("The difference among the groups is not statistically significant.")
```

The one-way ANOVA did not find a statistically significant difference in mean pTAU among the four sex and cognitive-status groups (F = 1.278, p = 0.288). Because the p-value was greater than 0.05, we failed to reject the null hypothesis. Therefore, this dataset did not provide sufficient evidence that mean pTAU differed among the four groups.
